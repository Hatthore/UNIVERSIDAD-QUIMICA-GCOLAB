```python
# ============================================
# CURVA DE CALIBRACIÓN PROTEÍNAS (BRADFORD)
# ============================================

# 1) Subir CSV(s)
from google.colab import files
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import io, textwrap

print("👉 Selecciona tu CSV de estándares (concentracion_mg_mL, absorbancia_595nm)")
uploaded = files.upload()
fname_std = list(uploaded.keys())[0]
df_std = pd.read_csv(io.BytesIO(uploaded[fname_std]))

# (Opcional) Subir CSV de muestras desconocidas
print("\n¿Quieres subir un CSV de muestras desconocidas ahora? (absorbancia_595nm)")
print("Si no, presiona 'cancelar' en el selector.")
try:
    up2 = files.upload()
    fname_unk = list(up2.keys())[0]
    df_unk = pd.read_csv(io.BytesIO(up2[fname_unk]))
except Exception:
    df_unk = None

# 2) Parámetros de análisis (ajústalos si quieres)
FORCE_THROUGH_ZERO = False    # True para forzar b = 0 (ajuste y = m*x)
BLANK_ABSORBANCE   = None      # p.ej., 0.050 si quieres restar blanco a TODAS las absorbancias

# 3) Limpieza y blanco
df_std = df_std.copy()
df_std.columns = [c.strip().lower() for c in df_std.columns]
assert "concentracion_mg_ml" in df_std.columns and "absorbancia_595nm" in df_std.columns, \
    "El CSV de estándares debe tener columnas 'concentracion_mg_mL' y 'absorbancia_595nm'."

x = df_std["concentracion_mg_ml"].astype(float).values
y = df_std["absorbancia_595nm"].astype(float).values

if BLANK_ABSORBANCE is not None:
    y = y - float(BLANK_ABSORBANCE)

# 4) Ajuste lineal
def linear_fit(x, y, force_zero=False):
    x = np.asarray(x, float); y = np.asarray(y, float)
    if force_zero:
        # m = sum(xy)/sum(x^2), b=0
        m = np.sum(x*y)/np.sum(x*x)
        b = 0.0
        y_pred = m*x
    else:
        # y = m*x + b (mínimos cuadrados)
        m, b = np.polyfit(x, y, 1)
        y_pred = m*x + b
    # R^2 clásico
    ss_res = np.sum((y - y_pred)**2)
    ss_tot = np.sum((y - np.mean(y))**2)
    r2 = 1 - ss_res/ss_tot if ss_tot > 0 else np.nan
    return m, b, y_pred, r2, ss_res, ss_tot

m, b, y_hat, r2, ss_res, ss_tot = linear_fit(x, y, FORCE_THROUGH_ZERO)

# 5) Figura de la curva (puntos + recta)
plt.figure(figsize=(7,5))
plt.scatter(x, y, label="Estándares", zorder=3)
# línea de ajuste extendida a todo el rango
xx = np.linspace(0, max(1e-9, x.max()*1.05), 100)
yy = m*xx + b
plt.plot(xx, yy, label="Ajuste lineal")
plt.xlabel("Concentración de proteína (mg/mL)")
plt.ylabel("Absorbancia a 595 nm")
titulo = "Curva de calibración Bradford"
if BLANK_ABSORBANCE is not None:
    titulo += " (abs restada)"
plt.title(titulo)
# Ecuación y R^2 en texto
eq = f"y = {m:.5f}·x + {b:.5f}" if not FORCE_THROUGH_ZERO else f"y = {m:.5f}·x"
txt = f"{eq}\nR² = {r2:.5f}"
plt.text(0.02*xx.max() if xx.max()>0 else 0.0, 0.85*max(yy.max(), y.max()+1e-9), txt,
         bbox=dict(boxstyle="round", facecolor="white", alpha=0.8))
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.savefig("curva_bradford.png", dpi=300, bbox_inches="tight")
plt.show()

# 6) Resumen estándares
resumen_std = pd.DataFrame({
    "concentracion_mg_mL": x,
    "absorbancia_595nm_proc": y,
    "predicho_por_ajuste": y_hat
}).sort_values("concentracion_mg_mL")

print("\n=== Resumen de ajuste (estándares) ===")
print(f"Ecuación: {eq}")
print(f"R²      : {r2:.5f}")
print(f"n puntos: {len(x)}")
display(resumen_std)

# 7) Cálculo de concentraciones para muestras desconocidas (si las subiste)
def concentration_from_abs(a, m, b, force_zero=False):
    if force_zero:
        return a / m
    else:
        return (a - b) / m

df_result_unk = None
if df_unk is not None:
    df_unk = df_unk.copy()
    df_unk.columns = [c.strip().lower() for c in df_unk.columns]
    assert "absorbancia_595nm" in df_unk.columns, "El CSV de muestras debe tener 'absorbancia_595nm'."
    A = df_unk["absorbancia_595nm"].astype(float).values
    if BLANK_ABSORBANCE is not None:
        A = A - float(BLANK_ABSORBANCE)
    conc = concentration_from_abs(A, m, b, FORCE_THROUGH_ZERO)
    df_result_unk = pd.DataFrame({
        "absorbancia_595nm_proc": A,
        "concentracion_mg_mL_calc": conc
    })
    print("\n=== Muestras desconocidas (calculadas) ===")
    display(df_result_unk)

# 8) Guardar salidas
out_coef = pd.DataFrame([{
    "slope_m": m,
    "intercept_b": b,
    "R2": r2,
    "FORCE_THROUGH_ZERO": FORCE_THROUGH_ZERO,
    "BLANK_ABSORBANCE": BLANK_ABSORBANCE
}])
out_coef.to_csv("bradford_ajuste_coeficientes.csv", index=False)
resumen_std.to_csv("bradford_estandares_resumen.csv", index=False)
if df_result_unk is not None:
    df_result_unk.to_csv("bradford_muestras_resultados.csv", index=False)

print("\nArchivos generados:")
print("- curva_bradford.png")
print("- bradford_ajuste_coeficientes.csv")
print("- bradford_estandares_resumen.csv")
if df_unk is not None:
    print("- bradford_muestras_resultados.csv")

# 9) Descargar resultados (si quieres)
from google.colab import files
try:
    files.download("curva_bradford.png")
    files.download("bradford_ajuste_coeficientes.csv")
    files.download("bradford_estandares_resumen.csv")
    if df_result_unk is not None:
        files.download("bradford_muestras_resultados.csv")
except:
    pass
```

```python
# ============================================
# Bradford: graficar ajustando SOLO con x>0
# (el 0 se usa solo como blanco para corregir A)
# ============================================
import numpy as np, pandas as pd, matplotlib.pyplot as plt

# === 1) Tomar X y A del DataFrame ===
x_all = df_std["concentracion_final_mg_mL"].astype(float).values
A_all = df_std["absorbancia_595nm"].astype(float).values

# === 2) Restar BLANCO (x = 0) y eliminarlo del ajuste ===
A_blank = float(A_all[x_all==0][0]) if (x_all==0).any() else 0.0
A_corr  = A_all - A_blank           # para mostrar y para muestras
mask    = x_all > 0                 # <-- aquí quitamos el 0 del ajuste
x, y    = x_all[mask], A_corr[mask]

# === 3) Ajuste lineal (elige una de las dos opciones) ===
FORCE_THROUGH_ZERO = True  # True -> y = m*x ; False -> y = m*x + b

def linear_fit(x, y, force_zero=False):
    x = np.asarray(x, float); y = np.asarray(y, float)
    if force_zero:
        m = (x@y)/(x@x); b = 0.0; yhat = m*x
    else:
        m, b = np.polyfit(x, y, 1); yhat = m*x + b
    ss_res = np.sum((y - yhat)**2)
    ss_tot = np.sum((y - y.mean())**2) if len(y)>1 else np.nan
    R2 = 1 - ss_res/ss_tot if ss_tot>0 else np.nan
    return m, b, yhat, R2

m, b, yhat, R2 = linear_fit(x, y, FORCE_THROUGH_ZERO)

# === 4) Gráfica ===
xx = np.linspace(0, max(1e-9, x_all.max()*1.05), 200)
yy = m*xx + b

plt.figure(figsize=(7,5))
plt.scatter(x, y, label="Estándares (x>0)", zorder=3)
plt.plot(xx, yy, label=f"Ajuste {'(b=0)' if FORCE_THROUGH_ZERO else ''}", zorder=2)
plt.xlabel("Concentración final (mg/mL)")
plt.ylabel("Absorbancia 595 nm (blanco restado)")
plt.title("Curva de calibración (sin el punto 0)")
eq = f"y = {m:.6f}·x" + ("" if FORCE_THROUGH_ZERO else f" + {b:.6f}")
plt.text(0.02*xx.max(), 0.90*max(yy.max(), y.max()+1e-9),
         f"{eq}\nR² = {R2:.5f}\nA_blank = {A_blank:.4f}",
         bbox=dict(facecolor="white", alpha=0.8))
plt.grid(True, alpha=0.3); plt.legend(); plt.tight_layout()
plt.savefig("curva_bradford_sin_cero.png", dpi=300, bbox_inches="tight")
plt.show()

print(f"Ecuación final: {eq} | R² = {R2:.5f}  (el 0 no se usó en la regresión)")
```
