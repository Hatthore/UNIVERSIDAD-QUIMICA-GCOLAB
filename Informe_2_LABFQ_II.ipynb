```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import linregress
```

```python
# Datos
w = np.array([0.000, 0.101, 0.205, 0.299, 0.400, 0.500, 0.600, 0.698, 0.794, 0.894, 1.000])
viscosidad = np.array([1.1117, 1.3415, 1.5144, 1.9811, 2.2379, 2.4157, 2.5514, 2.5592, 2.4540, 2.4498, 1.8064])

# Crear la gráfica
plt.figure(figsize=(8,6))
plt.plot(w, viscosidad, marker='o', linestyle='-', color='g')
# Agregar etiquetas a cada punto
for i, txt in enumerate(viscosidad):
    plt.text(w[i], viscosidad[i] + 0.05, f'{txt:.3f}', ha='center', fontsize=10)

# Etiquetas y título
plt.xlabel("Fracción Másica (adimensional)")
plt.ylabel("Viscosidad (mPa·s)")
plt.title("Viscosidad vs. Fracción másica")
plt.legend()
plt.grid()
plt.ylim(0, 3)


# Guardar la imagen en formato PNG
plt.savefig("Viscosidad vs. Fracción másica.png", dpi=300, bbox_inches="tight")

# Descargar la imagen
from google.colab import files
files.download("Viscosidad vs. Fracción másica.png")
```

```python
# Datos
fraccion_molar = [0.000, 0.032, 0.072, 0.113, 0.166, 0.231, 0.310, 0.410, 0.536, 0.716, 1.000]
viscosidad = ([1.1117, 1.3415, 1.5144, 1.9811, 2.2379, 2.4157, 2.5514, 2.5592, 2.4540, 2.4498, 1.8064])

# Gráfica
plt.figure(figsize=(8, 5))
plt.plot(fraccion_molar, viscosidad, marker='o', linestyle='-', color='r')

# Etiquetas
for i, txt in enumerate(viscosidad):
    plt.annotate(f'{txt:.2f}', (fraccion_molar[i], viscosidad[i]), textcoords="offset points", xytext=(0,5), ha='center')

plt.xlabel("Fracción Molar")
plt.ylabel("Viscosidad (mPa·s)")
plt.title("Viscosidad vs Fracción Molar")
plt.ylim(0, 3)
plt.legend()
plt.grid()


# Guardar la imagen en formato PNG
plt.savefig("Viscosidad vs Fracción Molar.png", dpi=300, bbox_inches="tight")

# Descargar la imagen
from google.colab import files
files.download("Viscosidad vs Fracción Molar.png")
```

```python
# Temperaturas en Kelvin
temperaturas = np.array([308.15, 318.15, 328.15, 333.15])

# Cálculo de 1/T
inv_temp = 1 / temperaturas

# Viscosidades
viscosidad_iso = np.array([1.4612859, 1.1886434, 1.0082277, 0.9594417])
viscosidad_agua = np.array([1.0341179, 1.0015005, 0.9075663, 0.9051653])

# Cálculo de ln(viscosidad)
ln_visco_iso = np.log(viscosidad_iso)
ln_visco_agua = np.log(viscosidad_agua)

# Ajuste lineal para ambas sustancias
slope_iso, intercept_iso, r_value_iso, _, _ = linregress(inv_temp, ln_visco_iso)
slope_agua, intercept_agua, r_value_agua, _, _ = linregress(inv_temp, ln_visco_agua)

# Generar valores para la recta
x_fit = np.linspace(inv_temp.min(), inv_temp.max(), 100)
y_fit_iso = slope_iso * x_fit + intercept_iso
y_fit_agua = slope_agua * x_fit + intercept_agua

# Gráfica
plt.figure(figsize=(8, 6))
plt.scatter(inv_temp, ln_visco_iso, color='blue', label='Isopropanol')
plt.plot(x_fit, y_fit_iso, '--', color='blue', label=f'Ajuste Isopropanol (Pendiente={slope_iso:.2f})')
plt.scatter(inv_temp, ln_visco_agua, color='red', label='Agua')
plt.plot(x_fit, y_fit_agua, '--', color='red', label=f'Ajuste Agua (Pendiente={slope_agua:.2f})')

# Etiquetas
plt.xlabel('$1/T$ (K$^{-1}$)')
plt.ylabel('$\ln(\eta)$(mPa·s)')
plt.title('Gráfico de $\ln(\eta)$ vs. $1/T$')
plt.legend()
plt.grid()

# Mostrar la gráfica
plt.show()
```
