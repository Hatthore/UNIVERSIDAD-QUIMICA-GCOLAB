```python
import pandas as pd
import matplotlib.pyplot as plt


from google.colab import files

uploaded = files.upload()  # Se abrirá un cuadro de diálogo para subir el archivo

df = pd.read_csv("negrokc1.csv", encoding="utf-8")  # Reemplaza con el nombre correcto del archivo
df.head()
df = df.dropna()  # Eliminar filas vacías
longitud_onda = df.iloc[:, 0]  # Primera columna
absorbancia = df.iloc[:, 1]    # Segunda columna
df["longitud de onda"] = pd.to_numeric(df["longitud de onda"], errors="coerce")
df["absorbancia"] = pd.to_numeric(df["absorbancia"], errors="coerce")
df = df.dropna()  # Eliminar datos nulos




plt.figure(figsize=(8, 5))
plt.plot(longitud_onda, absorbancia, label="Nanopartículas coloidales Negras", color="magenta")
plt.xlabel("Longitud de onda (nm)")
plt.ylabel("Absorbancia")
plt.title("Espectro UV-Vis de nanopartículas coloidades Negras")
plt.legend()
plt.grid()


# Guardar la imagen en formato PNG
plt.savefig("Espectro UV-Vis de nanopartículas coloidades Negras.png", dpi=300, bbox_inches="tight")

# Descargar la imagen
from google.colab import files
files.download("Espectro UV-Vis de nanopartículas coloidades Negras.png")
```
