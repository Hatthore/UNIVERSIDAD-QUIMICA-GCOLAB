```python
import pandas as pd
import matplotlib.pyplot as plt

from google.colab import files

uploaded = files.upload()  # Se abrirá un cuadro de diálogo para subir el archivo

# Ruta correcta de los archivos
archivo_negro = "/content/negrokc1.csv"
archivo_naranja = "/content/grish1.csv"

# Cargar datos
df_negro = pd.read_csv(archivo_negro)
df_naranja = pd.read_csv(archivo_naranja)

# Extraer datos (ajusta los nombres de columna si es necesario)
longitud_onda_negro = df_negro.iloc[:, 0]
absorbancia_negro = df_negro.iloc[:, 1]

longitud_onda_naranja = df_naranja.iloc[:, 0]
absorbancia_naranja = df_naranja.iloc[:, 1]

# Graficar ambos espectros
plt.figure(figsize=(8, 5))
plt.plot(longitud_onda_negro, absorbancia_negro, label="Partículas Negras", color="magenta")
plt.plot(longitud_onda_naranja, absorbancia_naranja, label="Partículas Naranjas", color="green")

# Configuración de la gráfica
plt.xlabel("Longitud de onda (nm)")
plt.ylabel("Absorbancia")
plt.title("Comparación de espectros UV-Vis")
plt.legend()
plt.grid()

# Mostrar la gráfica
plt.show()
```
