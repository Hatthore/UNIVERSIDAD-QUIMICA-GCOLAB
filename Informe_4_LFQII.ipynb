```python
import numpy as np

# Datos experimentales
masa_liquido = np.array([0.082, 0.073, 0.070, 0.098, 0.089, 0.079, 0.077, 0.076, 0.081, 0.082])  # en g
volumen_gas_ml = np.array([82, 87, 87.5, 84, 96, 85, 84, 83, 91, 97])  # en mL

# Constantes
R = 8.314  # J/mol·K
T = 369.9  # K
P = 9000   # Pa

# Conversión de volumen de mL a m^3
volumen_gas_m3 = volumen_gas_ml * 1e-6  # de mL a m^3

# Cálculo de moles usando la ecuación de estado de gases ideales: PV = nRT => n = PV / RT
n_moles = (P * volumen_gas_m3) / (R * T)

# Masa molar experimental = masa del líquido / moles del gas producido
masa_molar_exp = masa_liquido / n_moles  # en g/mol

# Resultados promedios
masa_molar_promedio = np.mean(masa_molar_exp)
desviacion_std = np.std(masa_molar_exp, ddof=1)

# Comparación con el valor ideal (masa molar teórica del metanol ≈ 32.04 g/mol)
masa_molar_ideal = 32.04
error_relativo = abs(masa_molar_promedio - masa_molar_ideal) / masa_molar_ideal * 100

# Impresión de resultados
print(f"Masa molar promedio experimental: {masa_molar_promedio:.2f} g/mol")
print(f"Desviación estándar: {desviacion_std:.2f} g/mol")
print(f"Error relativo respecto a la masa molar ideal: {error_relativo:.2f}%")
```

```python
import pandas as pd

# Datos experimentales
datos = {
    "Masa_jeringa_llena_g": [2.6, 2.605, 2.598, 2.616, 2.618, 2.604, 2.607, 2.603, 2.608, 2.609],
    "Masa_jeringa_vacia_g": [2.518, 2.532, 2.528, 2.518, 2.529, 2.525, 2.530, 2.527, 2.527, 2.527],
    "Volumen_vaporizado_mL": [82, 87, 87.5, 84, 96, 85, 84, 83, 91, 97]
}

# Constantes
R = 8.314  # J/mol·K
T = 369.9  # K
P = 90000 # Pa

# Cálculos
df = pd.DataFrame(datos)
df["Masa_liquido_g"] = df["Masa_jeringa_llena_g"] - df["Masa_jeringa_vacia_g"]
df["Volumen_m3"] = df["Volumen_vaporizado_mL"] * 1e-6  # mL a m³
df["Moles"] = (P * df["Volumen_m3"]) / (R * T)
df["Masa_molar"] = df["Masa_liquido_g"] / df["Moles"]

print(df[["Masa_liquido_g", "Volumen_vaporizado_mL", "Moles", "Masa_molar"]])
```

```python
import numpy as np
import pandas as pd

# Datos experimentales
masa_liquido = np.array([0.192, 0.198, 0.197, 0.206, 0.203, 0.199, 0.207, 0.190, 0.189, 0.199])  # g
volumen_vap = np.array([87, 81, 93, 93, 94, 92, 95, 88, 90, 93])  # mL
volumen_vap = volumen_vap / 1000  # convertir a L

# Constantes
R = 0.082057  # L·atm/(mol·K)
T = 369.9  # K
P = 90000 / 101325  # Pa a atm (1 atm = 101325 Pa)

# Constantes de Van der Waals para el metanol
a = 2.253  # L^2·atm/mol^2
b = 0.0431  # L/mol

# Masa molar ideal (calculada antes)
M_ideal = (masa_liquido * R * T) / (P * volumen_vap)

# Corrección de Van der Waals
correccion = masa_liquido * ((b - (a / (R * T))) / volumen_vap)

# Masa molar real corregida
M_real = M_ideal + correccion

# Resultados
df = pd.DataFrame({
    'Masa líquida (g)': masa_liquido,
    'Volumen gas (L)': volumen_vap,
    'M ideal (g/mol)': M_ideal,
    'Corrección': correccion,
    'M real (g/mol)': M_real
})

pd.set_option('display.precision', 3)
print(df)
```
